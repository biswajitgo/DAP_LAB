```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=pd.read_csv("Virat_Kohli_100s.csv" )
```

```python
df.head()
```

```python
df.isnull().sum()
```

```python
df.columns
```

```python
df['Position'] = df['Position'].astype(str)
df['Year']=df['Year'].astype(int)
```

```python
df.drop_duplicates(inplace=True)
```

```python
df['Captain'] = df['Captain'].map({
    'Yes': 1,
    'No': 0
})

df['Win'] = df['Win'].map({
    'Yes': 1,
    'No': 0
})

print(df[['Captain', 'Win']].corr())
```

```python

print("Mean")
print(df[['Score', 'Balls', 'Strike Rate']].mean())

print("\nMedian")
print(df[['Score', 'Balls', 'Strike Rate']].median())

print("\nMode")
print(df[['Score', 'Balls', 'Strike Rate']].mode())

print("\nStandard Deviation")
print(df[['Score', 'Balls', 'Strike Rate']].std())

print("\nVariance")
print(df[['Score', 'Balls', 'Strike Rate']].var())
```

```python
sns.set_style("whitegrid")
```

```python
plt.figure(figsize=(8, 5))
sns.countplot(
    data=df,
    x='Format',
    palette=['red', 'blue']
)
plt.title("Centuries by Match Format")
plt.savefig("format_dist.png")
plt.show()
```

```python
sns.countplot(data=df, x='Year', color='red',width=0.5)

plt.title("Centuries per Year")

plt.show()
```

```python
top_10 = df['Against'].value_counts().head(10).index

sns.countplot(
    data=df[df['Against'].isin(top_10)],
    x='Against',
    color='blue'
)

plt.xticks(rotation=90)

plt.title("Top 10 Opponents")

plt.show()
```

```python
sns.scatterplot(
    data=df,
    x='Score',
    y='Strike Rate',
    color='red'
)

plt.title("Strike Rate vs Score")

plt.show()
```

```python
sns.countplot(
    data=df,
    x='Win',
    color='blue'
)

plt.title("Match Result")

plt.show()
```

```python
format_counts = df['Format'].value_counts()

plt.pie(
    format_counts,
    labels=format_counts.index,
    autopct='%1.1f%%',
    colors=['red', 'blue', 'green']
)

plt.title("Centuries by Format")

plt.show()
```

```python
corr = df.corr(numeric_only=True)

sns.heatmap(corr, annot=True)

plt.title("Correlation Heatmap")

plt.show()
```

```python
sns.countplot(
    data=df,
    x='Captain',
    hue='Win'
)

plt.title("Captain vs Match Result")

plt.show()
```

```python
sns.countplot(
    data=df,
    x='Format',
    hue='Win'
)

plt.title("Win and Loss by Format")

plt.show()
```
